```python
import numpy as np
import ase
from ase import io
from sklearn.model_selection import train_test_split
import pickle
```

# Which similarity matrix do you want to pull the similarity values from?

```python
sim_matrix_path_500 = 'sim-matrix-500k/MBTR-RBF-1000000.0gamma-15000frames-sim-matrix.pkl'
sim_matrix_path_1500 = 'sim-matrix-1500k/MBTR-RBF-1000000.0gamma-15000frames-sim-matrix.pkl'
sim_matrix_path_2500 = 'sim-matrix-2500k/MBTR-RBF-1000000.0gamma-15000frames-sim-matrix.pkl'

aimd_path_500 = '/home/cganley2/ai4mat2025/submission2/mace-models/training-data/SiC_3C-aimd-500K.out'
aimd_path_1500 = '/home/cganley2/ai4mat2025/submission2/mace-models/training-data/SiC_3C-aimd-1500K.out'
aimd_path_2500 = '/home/cganley2/ai4mat2025/submission2/mace-models/training-data/SiC_3C-aimd-2500K.out'

n_frames_per_traj = 500

t_dict = {'500': [sim_matrix_path_500, aimd_path_500],
          '1500': [sim_matrix_path_1500, aimd_path_1500],
          '2500': [sim_matrix_path_2500, aimd_path_2500]}
```

```python
training_trajectory_3N = []
validation_trajectory_3N = []
test_trajectory_3N = []

for key in t_dict:
    with open(t_dict[key][0], 'rb') as file:
        sim_matrix = pickle.load(file)

    aimd_traj = ase.io.read(t_dict[key][1], index=':')

    size = sim_matrix.shape[0]
    lower_tri_indices = np.tril_indices(size, k=1)
    lower_tri_values = sim_matrix[lower_tri_indices]

    sorted_indices = np.argsort(lower_tri_values)
    sorted_coords = np.array([(lower_tri_indices[0][i], lower_tri_indices[1][i]) for i in sorted_indices])

    flat_sorted_coords = sorted_coords.flatten()

    unique_coords, indices = np.unique(flat_sorted_coords, return_index=True)
    first_instances = flat_sorted_coords[np.sort(indices)]
    training_trajectory_3N.append([aimd_traj[index] for index in first_instances[:n_frames_per_traj]])

    validation, test = train_test_split([aimd_traj[index] for index in range(len(aimd_traj)) if index not in first_instances[:n_frames_per_traj]], train_size=1200, random_state=42)
    validation_trajectory_3N.append(validation)
    test_trajectory_3N.append(test)

flat_training_trajectory_3N = sum(training_trajectory_3N, [])
flat_validation_trajectory_3N = sum(validation_trajectory_3N, [])
flat_test_trajectory_3N = sum(test_trajectory_3N, [])

ase.io.write('train-rbf-{0}frames.extxyz'.format(len(flat_training_trajectory_3N)), images=flat_training_trajectory_3N)
ase.io.write('validation-rbf-{0}frames.extxyz'.format(len(flat_validation_trajectory_3N)), images=flat_validation_trajectory_3N)
ase.io.write('test-rbf-{0}frames.extxyz'.format(len(flat_test_trajectory_3N)), images=flat_test_trajectory_3N)
```
