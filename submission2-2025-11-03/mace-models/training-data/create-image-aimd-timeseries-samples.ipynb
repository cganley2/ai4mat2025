```python
import numpy as np
import ase
from ase import io
import matplotlib.pyplot as plt
```

# RL Sampling of 3 AIMD Trajectories (500K, 1500K, 2500K)

```python
train_indices_rl_500K_500 = np.load('Diverse_Subset_from_500K/top_indices_rl_500K_500.npy')
train_indices_rl_1500K_500 = np.load('Diverse_Subset_from_1500K/top_indices_rl_1500K_500.npy')
train_indices_rl_2500K_500 = np.load('Diverse_Subset_from_2500K/top_indices_rl_2500K_500.npy')

sic_500k = ase.io.read('/home/cganley2/qe-data/sic-3c/simulations/aimd-500K/SiC_3C-aimd-500K.out', index=':')
sic_1500k = ase.io.read('/home/cganley2/qe-data/sic-3c/simulations/aimd-1500K/SiC_3C-aimd-1500K.out', index=':')
sic_2500k = ase.io.read('/home/cganley2/qe-data/sic-3c/simulations/aimd-2500K/SiC_3C-aimd-2500K.out', index=':')
```

```python
plt.rcParams.update({'font.size': 16})  # Set this to desired size

fig, ax = plt.subplots(1, 3, sharey=True, figsize=(12,8))

ax[0].plot(np.arange(len(sic_500k)), [structure.get_potential_energy() for index, structure in enumerate(sic_500k)])
ax[0].plot(train_indices_rl_500K_500, [sic_500k[index].get_potential_energy() for index in train_indices_rl_500K_500], 'r.', markersize=5)
ax[0].set_xlabel('500K', fontweight='bold')
ax[0].set_ylabel('Energy (eV)', fontweight='bold')
ax[0].text(0.5, 0.95, 'A', transform=ax[0].transAxes, fontweight='bold')

ax[1].plot(np.arange(len(sic_1500k)), [structure.get_potential_energy() for index, structure in enumerate(sic_1500k)])
ax[1].plot(train_indices_rl_1500K_500, [sic_1500k[index].get_potential_energy() for index in train_indices_rl_1500K_500], 'r.', markersize=5)
ax[1].set_xlabel('1500K', fontweight='bold')
ax[1].text(0.5, 0.95, 'B', transform=ax[1].transAxes, fontweight='bold')

ax[2].plot(np.arange(len(sic_2500k)), [structure.get_potential_energy() for index, structure in enumerate(sic_2500k)])
ax[2].plot(train_indices_rl_2500K_500, [sic_2500k[index].get_potential_energy() for index in train_indices_rl_2500K_500], 'r.', markersize=5)
ax[2].set_xlabel('2500K', fontweight='bold')
ax[2].text(0.5, 0.95, 'C', transform=ax[2].transAxes, fontweight='bold')

fig.suptitle('RL Sampling of AIMD Trajectories', fontweight='bold')
# fig.savefig('rl-sampling-aimd-trajectories.png', dpi=600)
```
