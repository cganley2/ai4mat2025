```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sic_500k = np.genfromtxt('travis-data/vacf_global-500k.csv', delimiter='; ')
sic_1500k = np.genfromtxt('travis-data/vacf_global-1500k.csv', delimiter='; ')
sic_2500k = np.genfromtxt('travis-data/vacf_global-2500k.csv', delimiter='; ')
```

```python
plt.rcParams.update({'font.size': 16})  # Set this to desired size

fig, ax = plt.subplots(3, 1, sharex=True, figsize=(12,8))

ax[0].plot(sic_500k[:, 0], sic_500k[:, 1])
# ax[0].set_xlabel('500K', fontweight='bold')
# ax[0].set_ylabel(r'Velocity ACF (pm$^{2}$/ps$^{2}$)', fontweight='bold')
ax[0].text(0.5, 0.85, 'A: 500K', transform=ax[0].transAxes, fontweight='bold')
ax[0].set_ylim([-1e8, 1e8])
ax[0].set_yticks([-1e8, 0, 1e8])

ax[1].plot(sic_1500k[:, 0], sic_1500k[:, 1])
# ax[0].set_xlabel('500K', fontweight='bold')
ax[1].set_ylabel(r'Velocity ACF (pm$^{2}$/ps$^{2}$)', fontweight='bold')
ax[1].text(0.5, 0.85, 'B: 1500K', transform=ax[1].transAxes, fontweight='bold')
ax[1].set_ylim([-1e8, 1e8])
ax[1].set_yticks([-1e8, 0, 1e8])

ax[2].plot(sic_2500k[:, 0], sic_2500k[:, 1])
# ax[0].set_xlabel('500K', fontweight='bold')
# ax[0].set_ylabel(r'Velocity ACF (pm$^{2}$/ps$^{2}$)', fontweight='bold')
ax[2].text(0.5, 0.85, 'C: 2500K', transform=ax[2].transAxes, fontweight='bold')
ax[2].set_ylim([-1e8, 1e8])
ax[2].set_yticks([-1e8, 0, 1e8])

fig.suptitle('VACF', fontweight='bold')
ax[2].set_xlabel('Correlation Depth (fs)', fontweight='bold')
# fig.savefig('correlation-time-aimd-trajectories.png', dpi=600)
```
