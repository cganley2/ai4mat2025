```python
import pickle
```

```python
# Open the pickle file in binary read mode
with open('energy-parity-plot.pkl', 'rb') as file:
    plotobj = pickle.load(file)

```

```python
fig, ax = plotobj
ax.tick_params(axis='both', which='major', labelsize=18)  # Example: set major ticks size
ax.set_xlabel('DFT', fontsize=20)
ax.set_ylabel('MACE Prediction', fontsize=20)
for annotation in ax.texts:
    annotation.set_size(20)  # set all annotation font sizes to 14
# ax.tick_params(axis='x', labelrotation=45)
ax.set_title('Energy Parity (eV)', fontsize=20)
ax.legend(fontsize=20)
fig.set_size_inches((12, 6))
# ax.set_aspect('equal')
fig.savefig('publication-energy-parity.png', dpi=300, bbox_inches='tight')
```
